```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set(style="whitegrid")
```

```python
df = pd.read_csv("penguins.csv")

df.head()
```

```python

print("Shape:", df.shape)


print("\nDataset Info:")
df.info()


print("\nSummary Statistics:")
print(df.describe())


print("\nMissing Values:")
print(df.isnull().sum())
```

```python

df = df.dropna()


print(df.isnull().sum())

#I handled the missing values by removing rows that contained missing data using dropna().
```

```python
plt.figure(figsize=(8,5))
sns.histplot(df["body_mass_g"], bins=20, color="skyblue")

plt.title("Distribution of Penguin Body Mass")
plt.xlabel("Body Mass (g)")
plt.ylabel("Count")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.scatterplot(
    data=df,
    x="bill_length_mm",
    y="body_mass_g",
    hue="species"
)

plt.title("Bill Length vs Body Mass")
plt.xlabel("Bill Length (mm)")
plt.ylabel("Body Mass (g)")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(
    data=df,
    x="species",
    y="flipper_length_mm"
)

plt.title("Flipper Length by Species")
plt.xlabel("Species")
plt.ylabel("Flipper Length (mm)")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.countplot(
    data=df,
    x="species"
)

plt.title("Number of Penguins by Species")
plt.xlabel("Species")
plt.ylabel("Count")

plt.show()
```
