```python
import pandas as pd

df = pd.read_csv("penguins.csv")
```

```python
df.info()
```

```python
print("Unique species:", df["species"].nunique())
print("Unique islands:", df["island"].nunique())
```

```python
df.groupby("species")["body_mass_g"].mean()
```

```python
df.groupby("island")["flipper_length_mm"].mean()
```

```python
df.groupby("species")["bill_length_mm"].agg(["min", "max", "mean"])
```

```python
df.groupby("island")["flipper_length_mm"].mean().sort_values(ascending=False)
```

```python
df.groupby("species").agg({
    "body_mass_g": "mean",
    "flipper_length_mm": "mean"
})
```

```python
df.groupby("island").agg({
    "body_mass_g": ["count", "mean", "std"]
})
```

```python
penguin_regions = pd.DataFrame({
    "island": ["Biscoe", "Dream", "Torgersen"],
    "region": ["Antarctica", "Antarctica", "Antarctica"]
})

merged_df = pd.merge(df, penguin_regions, on="island")

merged_df.head()
```

```python
merged_df.groupby("region")["body_mass_g"].mean()
```

```python
penguin_observations = pd.DataFrame({
    "observation_date": [
        "2019-01-15",
        "2019-02-20",
        "2020-01-05",
        "2020-03-18"
    ]
})

penguin_observations
```

```python
penguin_observations["observation_date"] = pd.to_datetime(
    penguin_observations["observation_date"]
)

penguin_observations["year"] = penguin_observations["observation_date"].dt.year

penguin_observations
```

```python
penguin_observations["key"] = range(len(penguin_observations))
df["key"] = range(len(df))

merged = pd.merge(df, penguin_observations, on="key", how="left")

merged.head()
```
